```python
# ============================================
# STEP 1: IMPORT NECESSARY LIBRARIES
# ============================================

# Pandas is used for reading and analyzing the dataset
import pandas as pd

# NumPy is used for numerical calculations
import numpy as np

# Matplotlib is used for creating graphs and plots
import matplotlib.pyplot as plt

# Seaborn is used for better-looking data visualizations
import seaborn as sns

# LinearRegression is used to build our Linear Regression model
from sklearn.linear_model import LinearRegression
```

```python
# ============================================
# READ THE UPLOADED DATASET
# ============================================

# Read the text file into a Pandas DataFrame
# sep=r'\s+' means the values are separated by spaces/tabs
data = pd.read_csv(
    "a_batch-DA project .txt",
    sep=r'\s+'
)

# Display the first 5 rows of the dataset
data.head()
```

```python
# ============================================
# STEP 3.1: CHECK DATASET SIZE
# ============================================

# Display the number of rows and columns
print("Number of rows and columns:", data.shape)
```

```python
# ============================================
# STEP 3.2: CHECK COLUMN NAMES
# ============================================

# Display the names of all columns
print("Column names:")
print(data.columns)
```

```python
# ============================================
# STEP 3.3: CHECK DATA TYPES
# ============================================

# Display information about columns and their data types
data.info()
```

```python
# ============================================
# STEP 3.4: CHECK FOR MISSING VALUES
# ============================================

# Count the number of missing values in each column
print("Missing values in each column:")
print(data.isnull().sum())
```

```python
# ============================================
# STEP 3.5: STATISTICAL SUMMARY
# ============================================

# Generate basic statistical information
# such as count, mean, standard deviation,
# minimum, maximum and quartile values
data.describe()
```

```python
# ============================================
# STEP 3.6: DISPLAY THE COMPLETE DATASET
# ============================================

# Display all rows of the dataset
data
```

```python
# ============================================
# STEP 4: DECLARE FEATURE AND TARGET VARIABLES
# ============================================

# X is the feature/input variable
# We use 'media' to predict sales
X = data[['media']]

# y is the target/output variable
# This is the value we want to predict
y = data['sales']

# Display the first 5 values of X
print("Feature variable (X):")
print(X.head())

# Display the first 5 values of y
print("\nTarget variable (y):")
print(y.head())
```

```python
# ============================================
# CHECK THE SHAPE OF X AND y
# ============================================

# Display the dimensions of the feature variable
print("Shape of X:", X.shape)

# Display the dimensions of the target variable
print("Shape of y:", y.shape)
```

```python
# ============================================
# STEP 5: PLOT SCATTER PLOT
# ============================================

# Create a scatter plot using media and sales
plt.scatter(X, y)

# Give a label to the X-axis
plt.xlabel("Media")

# Give a label to the Y-axis
plt.ylabel("Sales")

# Give a title to the graph
plt.title("Relationship Between Media and Sales")

# Display the graph
plt.show()
```

```python
# ============================================
# STEP 6: CHECK X AND y
# ============================================

# Check the data type of X
print("Data type of X:")
print(type(X))

# Check the data type of y
print("\nData type of y:")
print(type(y))

# Check the shape/dimensions of X
print("\nShape of X:")
print(X.shape)

# Check the shape/dimensions of y
print("\nShape of y:")
print(y.shape)

# Display the first 5 values of X
print("\nFirst 5 values of X:")
print(X.head())

# Display the first 5 values of y
print("\nFirst 5 values of y:")
print(y.head())
```

```python
# ============================================
# STEP 7: APPLY LINEAR REGRESSION MODEL
# ============================================

# Create a Linear Regression model
model = LinearRegression()

# Train (fit) the model using our feature X and target y
model.fit(X, y)

# Display a message to confirm that the model was trained
print("Linear Regression model trained successfully!")
```

```python
# ============================================
# FIND THE REGRESSION COEFFICIENT AND INTERCEPT
# ============================================

# Get the coefficient (slope) of the regression line
coefficient = model.coef_[0]

# Get the intercept of the regression line
intercept = model.intercept_

# Display the values
print("Regression Coefficient:", coefficient)
print("Regression Intercept:", intercept)
```

```python
# ============================================
# STEP 8: PREDICT SALES
# ============================================

# Use the trained model to predict Sales
# based on the Media values in X
y_pred = model.predict(X)

# Display the first 5 predicted Sales values
print("First 5 predicted Sales values:")
print(y_pred[:5])
```

```python
# ============================================
# STEP 9: PLOT THE REGRESSION LINE
# ============================================

# Plot the original actual data points
plt.scatter(X, y, label="Actual Data")

# Plot the regression line using predicted values
plt.plot(X, y_pred, label="Regression Line")

# Label the X-axis
plt.xlabel("Media")

# Label the Y-axis
plt.ylabel("Sales")

# Add a title to the graph
plt.title("Linear Regression: Media vs Sales")

# Display the legend
plt.legend()

# Display the graph
plt.show()
```

```python
# ============================================
# STEP 10: EVALUATE THE MODEL
# ============================================

# Import R-squared evaluation metric
from sklearn.metrics import r2_score

# Calculate the R-squared score
r2 = r2_score(y, y_pred)

# Display the R-squared score
print("R-squared (R²) Score:", r2)
```

```python
# ============================================
# CALCULATE MAE AND RMSE
# ============================================

# Import evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Calculate Mean Absolute Error
mae = mean_absolute_error(y, y_pred)

# Calculate Mean Squared Error
mse = mean_squared_error(y, y_pred)

# Calculate Root Mean Squared Error
rmse = np.sqrt(mse)

# Display the evaluation results
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
```

```python
# ============================================
# CALCULATE R-SQUARED SCORE
# ============================================

# Import the R-squared evaluation metric
from sklearn.metrics import r2_score

# Calculate R-squared using actual and predicted Sales
r2 = r2_score(y, y_pred)

# Display the R-squared score
print("R-squared (R²) Score:", r2)
```

```python
# ============================================
# STEP 12: MAKE A SAMPLE PREDICTION
# ============================================

# Set a Media value for which we want to predict Sales
media_value = 25

# Create a DataFrame with the same feature name
# that was used when training the model
new_data = pd.DataFrame({'media': [media_value]})

# Use the trained model to predict Sales
predicted_sales = model.predict(new_data)

# Display the prediction
print("Media value:", media_value)
print("Predicted Sales:", predicted_sales[0])
```
